```python
import pandas
#Calcular os indicadores(base de dados)
#Abrir a base de dados
caminho = "/home/raels97/Downloads/Vendas - Dez.xlsx" #Utilizar "r" antes da declaração do diretório, quando a estrutura do OS utilizar contra-barra "\"
base_dados = pandas.read_excel(caminho)

#Ver as informações da base de dados
display(base_dados)

#Somar faturamento de todos os produtos = somar a coluna de Valor Final
faturamento = base_dados["Valor Final"].sum()
#Somar a quantidade de produtos vendidos = somar a coluna quantidade
quantidade_produtos = base_dados["Quantidade"].sum()

print(faturamento)
print(quantidade_produtos)
```
